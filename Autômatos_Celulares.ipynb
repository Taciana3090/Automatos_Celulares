### **Importando as bibliotecas necessárias**

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
```

### **Funções**

```python
def converte_binario(numero):
    binario = bin(numero)[2:].zfill(8)
    return list(binario)

def gera_arte(numero_celulas, max_iteracoes, regra):
    g = np.zeros(numero_celulas)
    ng = np.zeros(numero_celulas)
    codigo = converte_binario(regra)
    matriz_evolucao = np.zeros((max_iteracoes, numero_celulas))
    g[len(g)//2] = 1
    for i in range(max_iteracoes):
        matriz_evolucao[i,:] = g
        for j in range(len(g)):
            esquerda = g[(j-1) % len(g)]
            centro = g[j]
            direita = g[(j+1) % len(g)]

            if esquerda == 0 and centro == 0 and direita == 0:
                ng[j] = int(codigo[7])
            elif esquerda == 0 and centro == 0 and direita == 1:
                ng[j] = int(codigo[6])
            elif esquerda == 0 and centro == 1 and direita == 0:
                ng[j] = int(codigo[5])
            elif esquerda == 0 and centro == 1 and direita == 1:
                ng[j] = int(codigo[4])
            elif esquerda == 1 and centro == 0 and direita == 0:
                ng[j] = int(codigo[3])
            elif esquerda == 1 and centro == 0 and direita == 1:
                ng[j] = int(codigo[2])
            elif esquerda == 1 and centro == 1 and direita == 0:
                ng[j] = int(codigo[1])
            elif esquerda == 1 and centro == 1 and direita == 1:
                ng[j] = int(codigo[0])
        g = ng.copy()
    return matriz_evolucao


def mostra_arte(matriz_evolucao):
    cmap = mcolors.ListedColormap(['#FF0000', '#00FF00', '#0000FF', '#FFFF00'])
    plt.imshow(matriz_evolucao, cmap=cmap, interpolation='nearest', aspect='auto')
    plt.axis('off')
    plt.show()
```

### **Parâmetros do autômato**

```python
# parâmetros do autômato
numero_celulas = 101
max_iteracoes = 63
regra = 30
matriz_evolucao = gera_arte(numero_celulas, max_iteracoes, regra)
mostra_arte(matriz_evolucao)
```
